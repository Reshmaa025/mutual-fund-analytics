```python
import numpy as np

def cagr(series, years):
    series = series.dropna()

    trading_days = years * 252

    if len(series) < trading_days:
        return np.nan

    start = series.iloc[-trading_days]
    end = series.iloc[-1]

    return (end/start)**(1/years) - 1
```

```python
import pandas as pd
import numpy as np

def cagr(series, years):
    series = series.dropna()

    if len(series) < years * 12:  # monthly data
        return np.nan

    start = series.iloc[-years*12]
    end = series.iloc[-1]

    return (end/start)**(1/years) - 1
```
